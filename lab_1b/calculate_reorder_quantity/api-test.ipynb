```python
import requests

# URL of your running FastAPI server
# url = "http://127.0.0.1:8000/calculate-reorder-quantity"
url = "https://calculate-reorder-quantity.1u1bpgsnsvuw.us-south.codeengine.appdomain.cloud/calculate-reorder-quantity"

# Payload to send
# payload = {
#     "current_inventory": 50,
#     "historic_data": 120,
#     "forecast": 100
# }
payload = {
    "current_inventory": 1000,
    "historic_data": 3400,
    "forecast": 3107
}

# Make the POST request
response = requests.post(url, json=payload)

# Print the response
if response.status_code == 200:
    print("Reorder Quantity:", response.json()["reorder_quantity"])
    print("Reasoning:", response.json()["reasoning"])
else:
    print("Request failed with status code:", response.status_code)
    print(response.text)
```
