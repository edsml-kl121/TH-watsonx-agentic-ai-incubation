```python
import requests
import json

# Base URL of your FastAPI server
# url = "http://127.0.0.1:8080/chat/completions"
url = "http://127.0.0.1:8080/chat/completions"

# Sample user message
user_query = "ช่วยค้นคว้าเกี่ยวกับผู้จัดจำหน่าย Xtralife ให้คะแนนซัพพลายเออร์จากบนลงล่างโดยพิจารณาจากตัวเลือกที่ดีที่สุดไปจนถึงแย่ที่สุด พร้อมทั้งแบ่งปันเหตุผลด้วย ข้อมูลราคาสำหรับซัพพลายเออร์ ทั้งหมด 3 เจ้า มีดังนี้: [{'Unit Price': 61.5, 'Pricebook Name': 'Excelentia Supplies'}, {'Unit Price': 76.9, 'Pricebook Name': 'Global Office Solutions'}, {'Unit Price': 92.3, 'Pricebook Name': 'CGV Supplier'}] พิจารณาข้อกำหนดและบทวิจารณ์การขายของซัพพลายเออร์เหล่านี้ด้วย"

# Payload for the POST request
payload = {
    "messages": [
        {"role": "user", "content": user_query}
    ],
    "stream": False,
    "extra_body": {
        "thread_id": "thread-123"
    }
}

# Headers including Authorization and optional thread ID
headers = {
    "Content-Type": "application/json",
    "Authorization": "Bearer API_KEY",  # Replace with actual token if needed
    "X-IBM-THREAD-ID": "thread-123"
}

# Send POST request
response = requests.post(url, headers=headers, data=json.dumps(payload))

# Check response status and print result
if response.status_code == 200:
    print("✅ Success")
    print("Response:")
    print(response.json())
else:
    print("❌ Failed with status code:", response.status_code)
    print("Response:", response.text)
```
