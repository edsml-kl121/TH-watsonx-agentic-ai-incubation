```python
import requests
import json

# Base URL of your FastAPI server
# url = "http://127.0.0.1:8000/checkup"  # Local development
url = "https://agentlab-th.1u1bpgsnsvuw.us-south.codeengine.appdomain.cloud"  # Production

# Test cases for different scenarios
test_cases = [
    {
        "name": "Test Case 1: Young person with normal BMI",
        "payload": {
            "bmi": 22.5,
            "age": 25
        }
    },
    {
        "name": "Test Case 2: Young person with high BMI", 
        "payload": {
            "bmi": 30.0,
            "age": 35
        }
    },
    {
        "name": "Test Case 3: Older person with normal BMI",
        "payload": {
            "bmi": 23.0,
            "age": 55
        }
    },
    {
        "name": "Test Case 4: Older person with high BMI",
        "payload": {
            "bmi": 28.5,
            "age": 65
        }
    },
    {
        "name": "Test Case 5: Edge case - BMI exactly 25, age exactly 40",
        "payload": {
            "bmi": 25.0,
            "age": 40
        }
    }
]

# Headers for the request
headers = {
    "Content-Type": "application/json",
    "Authorization": "Bearer API_KEY"  # Replace with actual token if needed
}

print("🏥 Testing Health Checkup Decision API")
print("=" * 60)

# Test each case
for i, test_case in enumerate(test_cases, 1):
    print(f"\n{i}. {test_case['name']}")
    print(f"   Input: BMI={test_case['payload']['bmi']}, Age={test_case['payload']['age']}")
    
    # Send POST request
    try:
        response = requests.post(url, headers=headers, data=json.dumps(test_case['payload']))
        
        # Check response status and print result
        if response.status_code == 200:
            result = response.json()
            print("   ✅ Success")
            print(f"   Plan: {result['plan']}")
            print(f"   Doctor: {result['doctor']}")
            print(f"   Decision: {result['decision']}")
            print(f"   Thai Message: {result['thai_message']}")
                
        else:
            print(f"   ❌ Failed with status code: {response.status_code}")
            print(f"   Response: {response.text}")
            
    except requests.exceptions.RequestException as e:
        print(f"   ❌ Request failed: {e}")

print("\n" + "=" * 60)

# Test invalid inputs
print("\n🚨 Testing Error Cases")
print("-" * 30)

error_test_cases = [
    {
        "name": "Invalid BMI (negative)",
        "payload": {"bmi": -5, "age": 30}
    },
    {
        "name": "Invalid age (too high)",
        "payload": {"bmi": 25, "age": 250}
    },
    {
        "name": "Missing BMI field",
        "payload": {"age": 30}
    }
]

for i, test_case in enumerate(error_test_cases, 1):
    print(f"\n{i}. {test_case['name']}")
    print(f"   Input: {test_case['payload']}")
    
    try:
        response = requests.post(url, headers=headers, data=json.dumps(test_case['payload']))
        
        if response.status_code == 400:
            print("   ✅ Correctly returned 400 Bad Request")
            error_detail = response.json().get('detail', 'No detail provided')
            print(f"   Error: {error_detail}")
        else:
            print(f"   ❌ Unexpected status code: {response.status_code}")
            print(f"   Response: {response.text}")
            
    except requests.exceptions.RequestException as e:
        print(f"   ❌ Request failed: {e}")

print("\n" + "=" * 60)

# Test other endpoints
print("\n🔍 Testing Other Endpoints")
print("-" * 30)

# Test root endpoint
try:
    root_url = url.replace('/checkup', '/')
    response = requests.get(root_url)
    if response.status_code == 200:
        print("\n✅ Root endpoint (/) working")
        print(f"   Response: {response.json()['message']}")
    else:
        print(f"\n❌ Root endpoint failed: {response.status_code}")
except Exception as e:
    print(f"\n❌ Root endpoint error: {e}")

# Test /test endpoint  
try:
    test_url = url.replace('/checkup', '/test')
    response = requests.get(test_url)
    if response.status_code == 200:
        print("\n✅ Test endpoint (/test) working")
        test_results = response.json()['test_results']
        print(f"   Found {len(test_results)} test results")
    else:
        print(f"\n❌ Test endpoint failed: {response.status_code}")
except Exception as e:
    print(f"\n❌ Test endpoint error: {e}")

print("\n🎉 Testing completed!")
```
